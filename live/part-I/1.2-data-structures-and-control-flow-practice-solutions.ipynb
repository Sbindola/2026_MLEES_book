## Exercise 1: Lists — build, modify, slice

```python
readings_celsius = [-2.3, -1.1, 0.4, 1.2, -0.8]
readings_celsius.append(-3.1)
readings_celsius.remove(0.4)
print(readings_celsius)              # [-2.3, -1.1, 1.2, -0.8, -3.1]
print(readings_celsius[0], readings_celsius[-1])
print(readings_celsius[1:3])         # [-1.1, 1.2]
print(len(readings_celsius))         # 5
```

## Exercise 2: Dictionaries — build, look up, remove

```python
records = [("JFJ", "Jungfraujoch", 3571), ("BAS", "Basel-Binningen", 316), ("LUG", "Lugano", 273)]

elevations_m = {}
for code, name, elevation_m in records:
    elevations_m[code] = elevation_m

print(elevations_m["JFJ"])                  # 3571
print(elevations_m.get("ZRH", "unknown"))   # unknown

del elevations_m["LUG"]
for code, elevation_m in elevations_m.items():
    print(code, elevation_m)
```

## Exercise 3: A pure function

```python
def mean(values):
    total = 0.0
    for v in values:
        total += v
    return total / len(values)

readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(round(mean(readings_celsius), 2))   # 6.88
```

## Exercise 4: A function with a default argument

```python
def is_warm(temp_celsius, threshold=10.0):
    if temp_celsius >= threshold:
        return True
    else:
        return False

print(is_warm(6.1))                     # False
print(is_warm(6.1, threshold=5.0))      # True
```

## Exercise 5: Pair two stations with zip

```python
dates = ["Mon", "Tue", "Wed", "Thu", "Fri"]
jfj_celsius = [-2.3, -1.1, 0.4, 1.2, -0.8]
bas_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]

for date, jfj, bas in zip(dates, jfj_celsius, bas_celsius):
    print(date, round(bas - jfj, 2))     # Mon 8.4, Tue 8.5, ...
```

## Exercise 6: Number the readings with enumerate

```python
week_celsius = [6.1, 7.4, 5.9, 8.2, 11.3, 9.8, 6.8]

warmest_celsius = week_celsius[0]     # start from the first reading, not from 0.0:
warmest_day = 1                       # a week of frost would never beat a 0.0 starting point

for index, temp_celsius in enumerate(week_celsius):
    day = index + 1                   # enumerate counts from 0; days are counted from 1
    print(f"day {day}: {temp_celsius} °C")
    if temp_celsius > warmest_celsius:
        warmest_celsius = temp_celsius
        warmest_day = day

print(f"warmest: {warmest_celsius} °C on day {warmest_day}")     # 11.3 °C on day 5
```

## Exercise 7: while, break, and continue

```python
mixed_celsius = [6.1, None, 5.9, None, 9.5, 6.8]

first_hot = None          # defined first, in case no hot reading is found
i = 0
while i < len(mixed_celsius):
    value = mixed_celsius[i]
    i += 1                            # advance before any continue, or the loop never ends
    if value is None:
        continue
    if value > 9.0:
        first_hot = value
        break

print("first hot reading:", first_hot)      # 9.5
```

## Exercise 8: Default arguments and returning two values

```python
def summarise(values, scale=1.0):
    total = 0.0
    for v in values:
        total += v * scale
    return total, total / len(values)

readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]

total_celsius, mean_celsius = summarise(readings_celsius)
print(round(total_celsius, 2), round(mean_celsius, 2))     # 34.4 6.88

print(round(summarise(readings_celsius, scale=2.0)[1], 2)) # 13.76

result = summarise(readings_celsius)
print(type(result), result)        # <class 'tuple'> — two values come back packed in a tuple
```

## Exercise 9: A variable number of arguments

```python
def warmest(*temps_celsius):
    best = temps_celsius[0]
    for t in temps_celsius:
        if t > best:
            best = t
    return best

print(warmest(6.1, 9.5, 5.9))                        # 9.5
readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(warmest(*readings_celsius))                     # 8.2
```

## Exercise 10: Fix the shared-default bug

```python
# out=[] is created once at definition and shared across calls, so results accumulate.
# the simplest fix is to drop the out parameter and build a fresh list inside.
def collect_above(values, threshold):
    out = []
    for v in values:
        if v > threshold:
            out.append(v)
    return out

first = collect_above([1.0, 5.0, 9.0], 4.0)
second = collect_above([2.0, 8.0], 4.0)
print(first)    # [5.0, 9.0]
print(second)   # [8.0], not contaminated
```

### Exercise 11: Document your function with a docstring

```python
def mean(values):
    """Return the arithmetic mean of a list of temperatures, in °C."""
    total = 0.0
    for v in values:
        total += v
    return total / len(values)

readings_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]
print(round(mean(readings_celsius), 2))    # 6.88
print(mean.__doc__)
help(mean)
```

### Exercise 12: Tuples — unpacking and immutability

```python
station_coords = (46.5475, 7.9853)

# 1. unpacking
lat, lon = station_coords
print(lat, lon)                       # 46.5475 7.9853

# 2. tuples are immutable
try:
    station_coords[0] = 47.0
except TypeError as err:
    print("TypeError:", err)          # 'tuple' object does not support item assignment

# 3. a coordinate is a fixed record: changing one half without the other would silently
#    produce a position that was never measured, so immutability is a safety feature
```

### Exercise 13: Means and a filter, with comprehensions

```python
stations = {
    "JFJ": {"readings_celsius": [-2.3, -1.1, 0.4, 1.2, -0.8]},
    "BAS": {"readings_celsius": [6.1, 7.4, 5.9, 8.2, 6.8]},
    "LUG": {"readings_celsius": [9.4, 10.2, 8.8, 11.1, 9.9]},
}
# reuse the function from Exercise 3 instead of summing inline
means = {code: mean(v["readings_celsius"]) for code, v in stations.items()}
cold_codes = [code for code, m in means.items() if m < 0.0]
print({code: round(m, 2) for code, m in means.items()})   # {'JFJ': -0.52, 'BAS': 6.88, 'LUG': 9.88}
print(cold_codes)                                          # ['JFJ']
```

### Exercise 14: Spot the impure function

```python
def drop_last_impure(readings):
    readings.pop()
    return readings
```

```python
original = [6.1, 7.4, 5.9]
drop_last_impure(original)
print(original)                    # [6.1, 7.4] — the caller's list was modified
```

```python
def drop_last_pure(readings):
    """Return a copy of the list without its last element."""
    trimmed = readings.copy()
    trimmed.pop()
    return trimmed
```

```python
original = [6.1, 7.4, 5.9]
shorter = drop_last_pure(original)
print(original, shorter)           # [6.1, 7.4, 5.9] [6.1, 7.4]

# both bugs come from the same fact: a function receives the caller's list itself, not a
# copy, so changing it in place changes the caller's data -- and a default list is one
# object that every call receives
```

### Exercise 15: Reject impossible temperatures

```python
def to_kelvin(temp_celsius: float) -> float:
    """Return a temperature converted from degrees Celsius to kelvin."""
    if temp_celsius < -273.15:
        raise ValueError(f"{temp_celsius} °C is below absolute zero")
    return temp_celsius + 273.15

for value in [20.0, -300.0, 0.0]:
    try:
        print(round(to_kelvin(value), 2))
    except ValueError as err:
        print("skipped:", err)
```
